```python
import pandas as pd
import unicodedata

# ============================================================
# NORMALIZAÇÃO
# ============================================================
def sem_acento(s):
    if pd.isna(s):
        return ""
    return unicodedata.normalize("NFKD", str(s)) \
                     .encode("ascii", "ignore").decode()

def norm_texto(s):
    return " ".join(sem_acento(s).split()).upper().strip()

def norm_ean(s):
    if pd.isna(s):
        return ""
    s = str(s).strip()
    if s.lower() in {"", "nan", "none", "null", "nat"}:
        return ""
    if s.endswith(".0"):
        s = s[:-2]
    # trata célula só com espaços ou caractere invisível
    if s.replace(" ", "").replace("\xa0", "") == "":
        return ""
    return s

# ============================================================
# LER
# ============================================================
dfA = pd.read_excel("CadastroPaulinho_30-09.xlsx",
                    sheet_name="DadosPDV", dtype=str, keep_default_na=False)
dfB = pd.read_excel("CadastroPdv.xlsx",
                    sheet_name="Planilha1", dtype=str, keep_default_na=False)

# ============================================================
# CHAVE + NORMALIZAÇÃO
# ============================================================
for df in (dfA, dfB):
    df["_ref"]   = df["Referencia"].apply(norm_texto)
    df["_cor"]   = df["Nome da Cor"].apply(norm_texto)
    df["_ean"]   = df["EAN"].apply(norm_ean)
    df["_chave"] = df["_ref"] + "|" + df["_cor"]

# ============================================================
# EAN DE B POR CHAVE
# ============================================================
# Só considera EANs NÃO vazios
b_por_chave = (
    dfB[dfB["_ean"] != ""]
    .groupby("_chave")["_ean"]
    .apply(lambda x: x.iloc[0])
    .to_dict()
)

# ============================================================
# COMPARAR
# ============================================================
resultado = []
for _, row in dfA.iterrows():
    chave  = row["_chave"]
    ean_a  = row["_ean"]
    ean_b  = b_por_chave.get(chave, "")   # "" se não achou a chave OU se EAN_B é vazio

    # ---- EAN_B vazio (chave não existe em B ou EAN_B está em branco)
    if ean_b == "":
        status = "Sem Ean criado"
        ean_d  = ""

    # ---- EAN_A bate com EAN_B
    elif ean_a == ean_b:
        status = "OK"
        ean_d  = ""

    # ---- EAN_A diferente do EAN_B
    else:
        status = "NAO ENCONTRADO EM B"
        ean_d  = ean_b

    resultado.append({
        "EAN":         ean_a,
        "Referencia":  row["Referencia"],
        "Nome da Cor": row["Nome da Cor"],
        "EAN_B":       ean_d,
        "Status":      status,
    })

res = pd.DataFrame(resultado)

# ============================================================
# RESUMO
# ============================================================
print("\n=== RESUMO POR STATUS ===")
print(res["Status"].value_counts().to_string())

# ============================================================
# EXPORTAR
# ============================================================
with pd.ExcelWriter("Resultado_Comparacao.xlsx", engine="openpyxl") as w:
    res.to_excel(w, sheet_name="Todos", index=False)
    res[res["Status"] == "OK"] \
        .to_excel(w, sheet_name="OK", index=False)
    res[res["Status"] == "Sem Ean criado"] \
        .to_excel(w, sheet_name="Sem Ean criado", index=False)
    res[res["Status"] == "NAO ENCONTRADO EM B"] \
        .to_excel(w, sheet_name="Nao Encontrado em B", index=False)

print("\nArquivo salvo: Resultado_Comparacao.xlsx")
```
